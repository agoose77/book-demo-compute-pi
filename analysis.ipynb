```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
px, py = np.random.uniform(low=-1, high=1,size=(2, 1_000_000))
```

```python
#| label: empty-plot
plt.figure()
plt.axis("equal")

# Circle
theta = np.linspace(0, 2 * np.pi, 100)
x = np.cos(theta)
y = np.sin(theta)
plt.plot(x, y)

# Square
x = np.r_[-1, 1, 1, -1, -1]
y = np.r_[-1, -1, 1, 1, -1]
plt.plot(x, y)

# Radius
plt.arrow(0, 0, 1, 0)
plt.text(0.5, 0.1, "r = 1")

plt.show()
```

```python
#| label: pi-result
# Compute PI
# A_{square} = l \times w = 2 \times 2 = 4
# A_{circle} = \pi r^2 = \pi (1)^2 = \pi
# Therefore \pi \approx 4 \times \frac{A_{circle}}{A_{square}}.
n_inside = 0
n = len(px)
for i in range(n):
    if px[i] ** 2 + py[i] ** 2 <= 1.0:
        n_inside += 1
pi = 4 * n_inside / n
print(f"Pi is approximately {pi:.4f}")
```

```python
#| label: scatter-plot
plt.figure()
plt.axis("equal")

# Square
x = np.r_[-1, 1, 1, -1, -1]
y = np.r_[-1, -1, 1, 1, -1]
plt.plot(x, y)

# Radius
plt.arrow(0, 0, 1, 0)
plt.text(0.5, 0.1, "r = 1")

# Scatter
every_n = 200
r = px**2 + py**2
plt.scatter(px[::every_n], py[::every_n], c=r[::every_n]<=1.0, s=1)

plt.show()
```
